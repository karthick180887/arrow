```python
from pyspark import SparkConf, SparkContext
from pyspark.streaming import StreamingContext
conf = (SparkConf().set("spark.cores.max", "2").set("spark.executor.memory", "1g"))
```

```python
sc = SparkContext.getOrCreate()
ssc = StreamingContext(sc, 5)            # 5-second batches
```

```python
lines = ssc.socketTextStream("jupyter_lab", 9999)   # not "localhost"
counts = lines.flatMap(lambda l: l.split()) \
              .map(lambda w: (w, 1)) \
              .reduceByKey(lambda a, b: a + b)
counts.pprint()
```

```python
ssc.start()
ssc.awaitTerminationOrTimeout(300)
```
